# PRODCOM Classification - Interactive Jupyter Notebook

This page provides access to the interactive **ClassifAI Notebook**, a
self-contained Jupyter notebook for classifying products according to
the PRODCOM classification system using OpenAI’s language models.

The notebook is designed to be: - **Self-contained**: No complex
dependencies (no Qdrant required) - **Easy to use**: Only OpenAI API Key
needed - **Fully documented**: Step-by-step instructions included -
**Ready for cloud**: Deployable on Onyxia/SSPCloud - **Reproducible**:
